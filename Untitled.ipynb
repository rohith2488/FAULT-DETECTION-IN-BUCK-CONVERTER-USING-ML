```python
# ===========================
# FAULT DETECTION IN BUCK CONVERTER USING ML
# ===========================

# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.utils import to_categorical

# ===========================
# STEP 1: DATASET GENERATION
# ===========================

# Set random seed for reproducibility
np.random.seed(42)

# Number of samples
num_samples = 5000

# Generate data for healthy conditions
Vin = np.random.uniform(22, 26, num_samples)  # Input voltage (22V - 26V)
Duty = np.random.uniform(0.3, 0.7, num_samples)  # Duty cycle (30% - 70%)
Vo_healthy = Vin * Duty  # Ideal output voltage
IL_healthy = np.random.uniform(0.5, 1.5, num_samples)  # Healthy inductor current
Io_healthy = Vo_healthy / np.random.uniform(5, 20, num_samples)  # Load current

# Generate fault conditions
fault_types = ['Switch_Open', 'Switch_Short', 'Cap_ESR_Increase', 'Cap_Degrade',
               'Inductor_Short', 'Core_Saturation', 'Overcurrent', 'Short_Circuit', 'No_Fault']

# Generate fault data (simulating variations due to fault)
fault_labels = np.random.choice(fault_types, num_samples)

Vo_faulty = Vo_healthy.copy()
IL_faulty = IL_healthy.copy()
Io_faulty = Io_healthy.copy()

# Add random variations to signals for fault conditions
for i in range(num_samples):
    if fault_labels[i] == 'Switch_Open':
        Vo_faulty[i] *= 0.1  # Drastically reduced voltage
        IL_faulty[i] *= 0.05
    elif fault_labels[i] == 'Switch_Short':
        Vo_faulty[i] = Vin[i]  # Full input voltage
        IL_faulty[i] *= 2
    elif fault_labels[i] == 'Cap_ESR_Increase':
        Vo_faulty[i] *= 0.95
        IL_faulty[i] *= 1.1
    elif fault_labels[i] == 'Cap_Degrade':
        Vo_faulty[i] *= 0.9
    elif fault_labels[i] == 'Inductor_Short':
        IL_faulty[i] *= 2
    elif fault_labels[i] == 'Core_Saturation':
        IL_faulty[i] *= 1.5
    elif fault_labels[i] == 'Overcurrent':
        Io_faulty[i] *= 2
    elif fault_labels[i] == 'Short_Circuit':
        Vo_faulty[i] = 0
        Io_faulty[i] = 0

# Create DataFrame
data = pd.DataFrame({
    'Vin': Vin,
    'Vo': Vo_faulty,
    'IL': IL_faulty,
    'Io': Io_faulty,
    'Duty': Duty,
    'Fault_Type': fault_labels
})

# Save dataset as CSV
data.to_csv("buck_converter_fault_data.csv", index=False)
print("✅ Dataset saved as 'buck_converter_fault_data.csv'.")



# ===========================
# STEP 2: DATA PREPROCESSING
# ===========================

# Load dataset
data = pd.read_csv('buck_converter_fault_data.csv')

# Split data into features and labels
X = data[['Vin', 'Vo', 'IL', 'Io', 'Duty']]
y = data['Fault_Type']

# Encode categorical labels
y_encoded, class_names = pd.factorize(y)

# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

# Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("✅ Data preprocessing completed.")

# ===========================
# STEP 3: RANDOM FOREST MODEL
# ===========================

# Train Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train_scaled, y_train)

# Evaluate Random Forest Model
rf_preds = rf_model.predict(X_test_scaled)
rf_acc = accuracy_score(y_test, rf_preds)
print("\n🌲 Random Forest Results:\n", classification_report(y_test, rf_preds, target_names=class_names))
print("Confusion Matrix:\n", confusion_matrix(y_test, rf_preds))
print(f"✅ Random Forest Accuracy: {rf_acc * 100:.2f}%")

# ===========================
# STEP 4: SVM MODEL
# ===========================

# Train SVM Classifier
svm_model = SVC(kernel='rbf', C=1.0, random_state=42)
svm_model.fit(X_train_scaled, y_train)

# Evaluate SVM Model
svm_preds = svm_model.predict(X_test_scaled)
svm_acc = accuracy_score(y_test, svm_preds)
print("\n✨ SVM Results:\n", classification_report(y_test, svm_preds, target_names=class_names))
print("Confusion Matrix:\n", confusion_matrix(y_test, svm_preds))
print(f"✅ SVM Accuracy: {svm_acc * 100:.2f}%")

# ===========================
# STEP 5: CNN MODEL
# ===========================

# Convert labels to one-hot encoding for CNN
y_train_cnn = to_categorical(y_train)
y_test_cnn = to_categorical(y_test)

# Define CNN model
cnn_model = Sequential()
cnn_model.add(Dense(64, input_shape=(X_train_scaled.shape[1],), activation='relu'))
cnn_model.add(Dropout(0.3))
cnn_model.add(Dense(32, activation='relu'))
cnn_model.add(Dense(y_train_cnn.shape[1], activation='softmax'))

cnn_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Train CNN model
cnn_history = cnn_model.fit(X_train_scaled, y_train_cnn, epochs=20, batch_size=32, validation_split=0.2)

# Evaluate CNN model
cnn_eval = cnn_model.evaluate(X_test_scaled, y_test_cnn)
cnn_acc = cnn_eval[1]
print(f"\n✅ CNN Accuracy: {cnn_acc * 100:.2f}%")

# ===========================
# STEP 6: MODEL COMPARISON
# ===========================

# Model accuracy comparison plot
model_names = ['Random Forest', 'SVM', 'CNN']
accuracies = [rf_acc * 100, svm_acc * 100, cnn_acc * 100]

plt.figure(figsize=(8, 5))
sns.barplot(x=model_names, y=accuracies, palette='viridis')
plt.title('Model Accuracy Comparison')
plt.ylabel('Accuracy (%)')
plt.show()

# ===========================
# STEP 7: CONFUSION MATRIX PLOTS
# ===========================

# Confusion matrix plots
def plot_confusion_matrix(model_name, y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
    plt.title(f'Confusion Matrix - {model_name}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()

plot_confusion_matrix('Random Forest', y_test, rf_preds)
plot_confusion_matrix('SVM', y_test, svm_preds)

# ===========================
# STEP 8: FEATURE IMPORTANCE (RANDOM FOREST)
# ===========================

# Plot feature importance for Random Forest
feature_importance = rf_model.feature_importances_
plt.figure(figsize=(8, 5))
sns.barplot(x=feature_importance, y=X.columns, palette='viridis')
plt.title('Feature Importance - Random Forest')
plt.xlabel('Importance Score')
plt.show()

# ===========================
# STEP 9: CNN TRAINING CURVES
# ===========================

# CNN Training Accuracy and Loss
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(cnn_history.history['accuracy'], label='Training Accuracy')
plt.plot(cnn_history.history['val_accuracy'], label='Validation Accuracy')
plt.title('CNN Accuracy Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(cnn_history.history['loss'], label='Training Loss')
plt.plot(cnn_history.history['val_loss'], label='Validation Loss')
plt.title('CNN Loss Over Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.show()

```
